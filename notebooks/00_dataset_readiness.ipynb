# 00 Dataset readiness

This notebook checks which public datasets are available locally and records how each one can support the paper. It does not unpack large archives. The output is a reproducibility table used to justify the empirical design choices.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().resolve()
while PROJECT_ROOT.name != "online_experiment_designs_under_interference" and PROJECT_ROOT != PROJECT_ROOT.parent:
    PROJECT_ROOT = PROJECT_ROOT.parent
sys.path.insert(0, str(PROJECT_ROOT / "code" / "src"))

from config import ProjectPaths
from data_readiness import dataset_readiness_table, kuairand_archive_members, kuairand_nested_members, open_bandit_archive_members
from progress import ProgressLogger
from tables import write_table

paths = ProjectPaths.from_anywhere(PROJECT_ROOT)
paths.ensure()
progress = ProgressLogger("00-readiness")
```

```python
progress.log("Scanning candidate datasets without unpacking large archives.")
readiness = dataset_readiness_table(paths.data_dir)
write_table(readiness, paths.table_dir / "00_dataset_readiness.csv")
readiness
```

```python
progress.log("Inspecting KuaiRand archive members for the recommended fast-start subsets.")
kuairand_zip = paths.data_dir / "KuaiRand" / "10439422.zip"
kuairand_members = kuairand_archive_members(kuairand_zip)
write_table(kuairand_members, paths.table_dir / "00_kuairand_archive_members.csv")
kuairand_members
```

```python
progress.log("Inspecting Open Bandit archive members for known-propensity datasets.")
open_bandit_zip = paths.data_dir / "open_bandit" / "open_bandit_dataset.zip"
open_bandit_members = open_bandit_archive_members(open_bandit_zip)
write_table(open_bandit_members, paths.table_dir / "00_open_bandit_members.csv")
open_bandit_members
```

```python
progress.log("Peeking inside KuaiRand-Pure to check available files.")
kuairand_pure_members = kuairand_nested_members(kuairand_zip)
write_table(kuairand_pure_members, paths.table_dir / "00_kuairand_pure_members.csv")
kuairand_pure_members.head(20)
```
